## Day 5（周五）函数

### 练习 7：把周二的密度计算改写成函数
预期结果：composite_density(2.5, 1.2, 0.4) 返回 1.72

```python
# TODO: 定义函数 composite_density(d_fiber, d_resin, vf)，返回理论密度
# 调用三次（换不同的纤维体积分数），打印结果
def composite_density(d_fiber, d_resin, vf):
    rho_c = d_fiber * vf + d_resin * (1 - vf)
    return rho_c
print(composite_density(2.5,1.2,0.4))
print(composite_density(2.5,1.2,0.5))
print(composite_density(2.5,1.2,0.6))
for i in range(3,8):
    vf = i/10
    rho_c = composite_density(2.5,1.2,vf)
    print(f"vf = {vf}: 密度 = {rho_c:.2f}g/cm\u00b3")
densities = []
for i in range(3,8):
    vf = i/10
    rho=composite_density(2.5, 1.2, vf)
    densities .append(rho)
print(densities)
print(f"最大密度: {max(densities):.2f}")
print(f"最小密度: {min(densities):.2f}")
```
